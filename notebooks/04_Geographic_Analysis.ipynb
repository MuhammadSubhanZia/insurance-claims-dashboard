# Geographic Analysis

## Objective

Analyze recovery performance across different accident locations to identify regions with the highest financial exposure, collection efficiency, and operational opportunities.

### Business Questions

- Which locations have the highest number of claims?
- Which locations contribute the highest recovery amount?
- Which locations have the highest outstanding balance?
- Which locations collect money most efficiently?
- Which locations require management attention?

```python
import pandas as pd
import numpy as np
import plotly.express as px
from src.preprocessing import prepare_data

pd.set_option("display.max_columns", None) 
```

```python
df= prepare_data()
```

```python
location_summary = (
    df.groupby("Location")
      .agg(
          Claims=("Claim ID", "count"),
          Recovery=("Recovery Amount", "sum"),
          Collected=("Collected Amount", "sum"),
          Remaining=("Remaining Amount", "sum")
      )
      .reset_index()
)

location_summary["Collection Rate"] = (
    location_summary["Collected"] /
    location_summary["Recovery"] * 100
).round(2)

location_summary["Avg Recovery per Claim"] = (
    location_summary["Recovery"] /
    location_summary["Claims"]
).round(2)

location_summary.sort_values(
    "Recovery",
    ascending=False,
    inplace=True
)

location_summary.head(10)
```

### Q1) Which locations generate the highest number of claims?

```python
fig = px.bar(
    location_summary.sort_values("Claims").tail(10),
    x="Claims",
    y="Location",
    orientation="h",
    color="Claims",
    text_auto=".2s",
    title="Claims by Accident Location"
)

fig.show()
```

### Insight
This visualization identifies accident hotspots and regions contributing the highest operational workload.

### Q2) Which locations have the highest recovery portfolio?

```python
fig = px.bar(
    location_summary.sort_values("Recovery").tail(10),
    x="Recovery",
    y="Location",
    orientation="h",
    color="Recovery",
    text_auto=".2s",
    title="Recovery by Accident Location"
)

fig.show()
```

### Insight
Locations with the highest recovery amounts represent the largest financial exposure and should receive additional monitoring.

### Q3) Which locations have the highest outstanding balance?

```python
fig = px.bar(
    location_summary.sort_values("Remaining").tail(10),
    x="Remaining",
    y="Location",
    orientation="h",
    color="Remaining",
    text_auto=".2s",
    title="Outstanding Recovery by Location"
)

fig.show()
```

### Insight
High outstanding balances indicate locations where recovery efforts should be prioritized.

### Q4) Which locations recover money most efficiently?

```python
fig = px.bar(
    location_summary.sort_values("Collection Rate").tail(10),
    x="Collection Rate",
    y="Location",
    orientation="h",
    color="Collection Rate",
    text="Collection Rate",
    title="Collection Rate by Location (%)",
    color_continuous_scale="Viridis"
)

fig.show()
```

### Insight
Collection Rate normalizes performance across locations and highlights regions with the most effective recovery process.

### Q5) Average Recovery per Claim

```python
fig = px.bar(
    location_summary.sort_values("Avg Recovery per Claim").tail(10),
    x="Avg Recovery per Claim",
    y="Location",
    orientation="h",
    color="Avg Recovery per Claim",
    text_auto=".2s",
    title="Average Recovery Amount per Claim"
)

fig.show()
```

### Insight
This metric highlights locations where individual claims tend to be financially larger, helping distinguish claim volume from claim severity.

### Geographic Scorecard

```python
scorecard = location_summary[
    [
        "Location",
        "Claims",
        "Recovery",
        "Collected",
        "Remaining",
        "Collection Rate",
        "Avg Recovery per Claim"
    ]
].head(10)

scorecard
```

### Executive Insights

```python
top_claims = location_summary.loc[
    location_summary["Claims"].idxmax()
]

top_recovery = location_summary.loc[
    location_summary["Recovery"].idxmax()
]

top_collection = location_summary.loc[
    location_summary["Collection Rate"].idxmax()
]

top_remaining = location_summary.loc[
    location_summary["Remaining"].idxmax()
]

print(f"""
📍 EXECUTIVE GEOGRAPHIC INSIGHTS

• Highest Claim Volume:
  {top_claims['Location']} ({top_claims['Claims']} claims)

• Largest Recovery Portfolio:
  {top_recovery['Location']}
  SAR {top_recovery['Recovery']:,.0f}

• Highest Collection Efficiency:
  {top_collection['Location']}
  ({top_collection['Collection Rate']:.2f}%)

• Largest Outstanding Balance:
  {top_remaining['Location']}
  SAR {top_remaining['Remaining']:,.0f}
""")
```

### Locations by Bussiness Priority

```python
conditions = [
    location_summary["Collection Rate"] >= 70,
    location_summary["Collection Rate"].between(40, 70),
    location_summary["Collection Rate"] < 40
]

choices = [
    "🟢 Healthy",
    "🟡 Monitor",
    "🔴 Critical"
]

location_summary["Priority"] = np.select(
    conditions,
    choices,
    default="Unknown"
)

location_summary[
    [
        "Location",
        "Collection Rate",
        "Priority"
    ]
].sort_values("Collection Rate").sample(10) #run it multiple times to see different samples
```

```python
fig = px.scatter(
    location_summary,
    x="Recovery",
    y="Collection Rate",
    size="Claims",
    color="Remaining",
    hover_name="Location",
    title="Location Performance Matrix",
    labels={
        "Recovery": "Recovery Amount",
        "Collection Rate": "Collection Rate (%)"
    }
)

fig.show()
```
